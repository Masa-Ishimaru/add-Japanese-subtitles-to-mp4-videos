## 文字起こし

```python
audio_files = glob.glob('./input_audio/*.mp3')
audio_files = [file.replace('\\', '/') for file in audio_files]
audio_files
```

```python
model = whisper.load_model('large')
```

```python
for audio_file in audio_files:
    clear_output()
    audio_name = extract_video_name(audio_file)
    
    # オーディオファイル情報の表示
    sound_input = AudioSegment.from_file(audio_file, format = "mp3")
    sound_duration = sound_input.duration_seconds
    print('audio name = ', audio_name, ' audio duration = ', round(sound_duration / 60, 2), 'min *** Transcription starting...\n')    

    # 文字起こし
    result  = model.transcribe(audio_file, verbose = True, language = "en")
    seginfo = result["segments"]
    out_text= []

    # segment情報から発言の開始/終了時間とテキストを抜き出し、srt形式で編集する
    for data in seginfo:
        start = data["start"]
        end   = data["end"]
        text  = data["text"]
        out_line = Subtitle(index = 1,
                    start = timedelta(seconds = timedelta(seconds = start).seconds,
                    microseconds = timedelta(seconds = start).microseconds),
                    end = timedelta(seconds = timedelta(seconds = end).seconds,
                    microseconds = timedelta(seconds = end).microseconds),
                    content = text,
                    proprietary = '')
        out_text.append(out_line)
        
        
    # srt形式のファイルをcsv形式に編集して保存する。
    with open("csv_files/" + audio_name + ".csv", mode = "w", encoding = "utf-8_sig") as f:
        origin = srt.compose(out_text)
        origin = origin.replace(",", ".")
        origin = origin.replace("\n", ",")
        origin = origin.replace(",,", "\n")
        f.write(origin)
```

## 字幕テキストデータは読みやすいように少ない文字で改行する

```python
# 指定の文字数で改行する関数

def change_lines(text_raw, num_letters):
    str_dict = {}
    num_str_chunk = len(text_raw) // num_letters
    for i in range(num_str_chunk + 1):
        str_dict[i] = text_raw[num_letters * i : num_letters * (i + 1)]
    
    new_text = ''
    for i in range(num_str_chunk + 1):
        new_text = new_text + str_dict[i] + '\n'
    new_text = new_text[:-1]
    
    return new_text
```

## srt ファイルから ass ファイルを作成し、ass フォルダに保存する

```python
# srtファイルのパスの所得

files = glob.glob('./srt_files_ja/*.srt')
files = [file.replace('\\', '/') for file in files]
files
```

```python
# assファイルの作成

def convert_srt2ass(file_path):
    video_name = extract_video_name(file_path)
    print('processing ass file ', video_name, '...')
    
    # assファイルの作成
    new_path = shutil.copy('./srt_files_ja/' + video_name + '.srt', 'temp.srt')
    cmd = "ffmpeg -i temp.srt temp.ass"
    res = subprocess.call(cmd, shell = True)
    
    # assファイルを保存
    new_path = shutil.move('temp.ass', 'ass_files_ja/' + video_name + '_.ass')
```

```python
for file in files:
    convert_srt2ass(file)
```

## フォントサイズを読みやすいように変更する

```python
# フォントサイズを変更する関数

def replace_words_in_ass(file_path):
    video_name = extract_video_name(file_path)
    print('processing fontsize ', video_name, '...')

    with open('./ass_files_ja/' + video_name + '_.ass', encoding = "utf-8") as f:
        ass_file = f.read()
        
        pre  = 'Style: Default,Arial,16'     # 元のフォントサイズ = 16
        post = 'Style: Default,Arial,24'     # 変更したフォントのサイズ

    ass_file = ass_file.replace(pre, post)

    with open('./ass_files_ja/' + video_name + '_.ass', mode = "w", encoding = "utf-8") as f:
        f.write(ass_file)
```

```python
# assファイルを作成しフォントサイズ変更を適用

for file in files:
    convert_srt2ass(file)
    replace_words_in_ass(file)
```
